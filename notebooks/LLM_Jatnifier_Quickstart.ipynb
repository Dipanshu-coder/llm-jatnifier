## 2. Choose your model provider

Set `PROVIDER` to `huggingface` or `nvidia`. Create a token in that provider's account, then paste it when prompted. The token is requested privately and is not saved in this notebook.

```python
from getpass import getpass
from llm_output_polisher import Polisher

PROVIDER = "huggingface"  # change to "nvidia" if preferred

if PROVIDER == "huggingface":
    BASE_URL = "https://router.huggingface.co/v1"
    MODEL = "Qwen/Qwen2.5-7B-Instruct"
    API_KEY = getpass("Hugging Face token (hf_...): ")
elif PROVIDER == "nvidia":
    BASE_URL = "https://integrate.api.nvidia.com/v1"
    MODEL = "meta/llama-3.1-8b-instruct"
    API_KEY = getpass("NVIDIA API key: ")
else:
    raise ValueError("PROVIDER must be 'huggingface' or 'nvidia'")

polisher = Polisher(base_url=BASE_URL, model=MODEL, api_key=API_KEY)
print(f"Ready: {PROVIDER} / {MODEL}")
```

## 3. Polish any model output

Paste text from any LLM into `raw_output`. Jatnifier makes a second model call to improve presentation; it does not verify facts.

```python
raw_output = "photosynthesis is process plants use sunlight and co2 water to make glucose oxygen"

polished = polisher.polish(raw_output, profile="study", raise_errors=True)
print(polished)
```

## Optional: change the style

Available profiles: `general`, `study`, `technical`, `concise`. For a free local cleanup with no API call, use `Polisher.format_only(raw_output)`.

```python
print(polisher.polish(raw_output, profile="concise", raise_errors=True))

# No model/API call:
print(Polisher.format_only("###heading\n\n\n• first point"))
```
