## Choose provider and enter its API key
Set `PROVIDER` to `huggingface` or `nvidia`. Your key is entered privately and is not written into this notebook cell.

```python
from getpass import getpass
from llm_output_polisher import polish

PROVIDER = "huggingface"  # or "nvidia"
key = getpass("API key: ")
raw = "photosynthesis is process plants use sunlight and co2 water to make glucose oxygen"

print(polish(raw, provider=PROVIDER, api_key=key, style="study"))
```
